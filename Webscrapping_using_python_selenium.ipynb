```python
#month data
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import pandas as pd
import time

# Set up the Chrome WebDriver
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

# Open the target webpage
url = 'https://www.timeanddate.com/weather/india/hyderabad/historic'
driver.get(url)

# Select the month dropdown

month_dropdown = Select(driver.find_element(By.NAME, 'month'))
month_dropdown.select_by_visible_text('May 2024')

# Fetch data for each day in May
all_data = []

for day in range(1, 32):  # May has 31 days
    # Select the day
    day_dropdown = Select(driver.find_element(By.NAME, 'start'))
    day_dropdown.select_by_visible_text(f'{day} May 2024')
    
    # Wait for the page to load
    time.sleep(2)
    
    # Get the page source
    html = driver.page_source

    # Parse the HTML with BeautifulSoup
    soup = BeautifulSoup(html, 'html.parser')

    # Find all tables in the HTML
    tables = soup.find_all('table')

    # # Debug: Print the number of tables found
    # print(f"Number of tables found: {len(tables)}")

    # Extract and collect the content of each table, except the first one
    for index, table in enumerate(tables):
        if index == 0:
            continue  # Skip the first table
        
        # print(f"Table {index}:")

        # Extract all rows, including headers
        rows = []
        for row in table.find_all('tr'):
            cells = row.find_all(['th', 'td'])
            cell_text = [cell.get_text(strip=True) for cell in cells]
            rows.append(cell_text)
        
        if rows:
            rows.pop(0)  # Remove the header row
            rows.pop(-1)  # Remove the last row if it's not needed

        # Collect rows for all days
        all_data.extend(rows)

# Convert all collected data to DataFrame
data = pd.DataFrame(all_data)
data = data.reset_index(drop=True)
print(data)

# Close the browser
driver.quit()
```

```python
#yearly data
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import pandas as pd
import calendar
import time

# Set up the Chrome WebDriver
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

# Open the target webpage
url = 'https://www.timeanddate.com/weather/india/hyderabad/historic'
driver.get(url)

# Function to fetch data for a specific day
def fetch_day_data(month, day):
    try:
        day_dropdown = Select(driver.find_element(By.NAME, 'start'))
        day_dropdown.select_by_visible_text(f'{day} {calendar.month_name[month]} 2024')
        
        # Wait for the page to load
        time.sleep(2)
        
        # Get the page source
        html = driver.page_source

        # Parse the HTML with BeautifulSoup
        soup = BeautifulSoup(html, 'html.parser')

        # Find all tables in the HTML
        tables = soup.find_all('table')

        all_data = []
        
        for index, table in enumerate(tables):
            if index == 0:
                continue  # Skip the first table
            
            # Extract all rows, including headers
            rows = []
            for row in table.find_all('tr'):
                cells = row.find_all(['th', 'td'])
                cell_text = [cell.get_text(strip=True) for cell in cells]
                rows.append(cell_text)
            
            if rows:
                rows.pop(0)  # Remove the header row
                rows.pop(-1)  # Remove the last row if it's not needed
            
            all_data.extend(rows)
        
        return all_data

    except Exception as e:
        print(f"Error fetching data for {day} {calendar.month_name[month]} 2024: {e}")
        return []

# Fetch data for the entire year
all_year_data = []

for month in range(1, 13):  # Iterate over all months
    month_dropdown = Select(driver.find_element(By.NAME, 'month'))
    month_dropdown.select_by_visible_text(f'{calendar.month_name[month]} 2024')
    
    days_in_month = calendar.monthrange(2024, month)[1]
    
    for day in range(1, days_in_month + 1):
        day_data = fetch_day_data(month, day)
        all_year_data.extend(day_data)

# Convert all collected data to DataFrame
data = pd.DataFrame(all_year_data)
data = data.reset_index(drop=True)
print(data)


# Close the browser
driver.quit()
```

```python
#multiple month

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import pandas as pd
import calendar
import time

# Set up the Chrome WebDriver
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

# Open the target webpage
url = 'https://www.timeanddate.com/weather/india/hyderabad/historic'
driver.get(url)

# Function to fetch data for a specific day
def fetch_day_data(month, day):
    day_dropdown = Select(driver.find_element(By.NAME, 'start'))
    day_dropdown.select_by_visible_text(f'{day} {calendar.month_name[month]} 2024')
    
    # Wait for the page to load
    time.sleep(2)
    
    # Get the page source
    html = driver.page_source

    # Parse the HTML with BeautifulSoup
    soup = BeautifulSoup(html, 'html.parser')

    # Find all tables in the HTML
    tables = soup.find_all('table')

    all_data = []
    
    for index, table in enumerate(tables):
        if index == 0:
            continue  # Skip the first table
        
        # Extract all rows, including headers
        rows = []
        for row in table.find_all('tr'):
            cells = row.find_all(['th', 'td'])
            cell_text = [cell.get_text(strip=True) for cell in cells]
            rows.append(cell_text)
        
        if rows:
            rows.pop(0)  # Remove the header row
            rows.pop(-1)  # Remove the last row if it's not needed
        
        all_data.extend(rows)
    
    return all_data

# Function to fetch data for a specific month
def fetch_month_data(month):
    month_dropdown = Select(driver.find_element(By.NAME, 'month'))
    month_dropdown.select_by_visible_text(f'{calendar.month_name[month]} 2024')
    
    days_in_month = calendar.monthrange(2024, month)[1]
    
    all_month_data = []
    for day in range(1, days_in_month + 1):
        day_data = fetch_day_data(month, day)
        all_month_data.extend(day_data)
    
    return all_month_data

# Fetch data for multiple months
months_to_fetch = [5, 6]  # May and June
all_year_data = []

for month in months_to_fetch:
    month_data = fetch_month_data(month)
    all_year_data.extend(month_data)

# Convert all collected data to DataFrame
data = pd.DataFrame(all_year_data)
data = data.reset_index(drop=True)
print(data)

# Close the browser
driver.quit()
```
